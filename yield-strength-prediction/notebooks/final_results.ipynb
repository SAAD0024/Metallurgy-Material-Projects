# Steel Yield Strength Prediction — Final Results

Materials-engineering machine-learning results recorded from the completed analysis. The raw source database is not committed to the repository.


```python
import pandas as pd
import matplotlib.pyplot as plt

results = pd.DataFrame({
    'Model': ['Core Random Forest', 'Temperature-Augmented Random Forest'],
    'MAE (MPa)': [91.239146, 89.027748],
    'RMSE (MPa)': [139.367916, 138.460858],
    'R2': [0.796630, 0.799269]
})
display(results.round(4))
```

## Leakage-Free 5-Fold Cross-Validation

Temperature-augmented pipeline: **MAE 91.27 ± 2.25 MPa**, **RMSE 139.61 ± 3.61 MPa**, **R² 0.7878 ± 0.0181**.


```python
cv = pd.DataFrame({
    'Fold': [1,2,3,4,5],
    'MAE (MPa)': [89.493835,94.104120,93.110635,91.654343,88.020279],
    'RMSE (MPa)': [138.751687,138.616024,146.568841,136.050753,138.038013],
    'R2': [0.798425,0.778848,0.759961,0.788222,0.813656]
})
display(cv.round(4))
```

```python
ax = results.set_index('Model')[['MAE (MPa)','RMSE (MPa)']].plot(kind='bar', figsize=(9,6))
ax.set_ylabel('Error (MPa)')
ax.set_title('Core vs Temperature-Augmented Random Forest')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
ax = results.set_index('Model')[['R2']].plot(kind='bar', figsize=(8,6))
ax.set_ylabel('R²')
ax.set_title('R² Comparison')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Interpretation

- Temperature features reduced holdout MAE by **2.21 MPa (2.42%)** and RMSE by **0.91 MPa (0.65%)**.
- R² increased by **0.0026 (0.33%)**.
- Strong final features included `is_tempered`, `Fe`, `P`, `is_aged`, `Mo`, `Mn`, and `C`.
- Error becomes less consistent for sparsely represented high-strength materials.

> **Validation note:** the original hyperparameter search used the full modeling dataset before the reported holdout evaluation. The leakage-free 5-fold pipeline is therefore the primary robustness estimate; the holdout is a comparative evaluation rather than a completely untouched final test set.
